```python
import pandas as pd
import json
import csv
```

```python
# File di input e output
csv_file = "fewShot_output.csv"

# Apri il CSV
with open(csv_file, "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)

    # Intestazione
    writer.writerow(["ID", "Polarity", "Text"])
    id = 0

    # Carica il JSON
    emotions = ["positive", "neutral", "negative"]
    for emotion in emotions:
        json_file = "fewShot_generation_Ollama_llama3.2_1b_" + emotion + ".json"
        with open(json_file, "r", encoding="utf-8") as f:
            data = json.load(f)
            
            for item in data:
                id = id + 1
                emotion = item.get("target_emotion")
                text = item.get("result", {}).get("Text")
                
                writer.writerow([id, emotion, text])
```
